```python
import os
import math
import json
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch.cuda.amp import autocast, GradScaler
from datasets import load_dataset
from tqdm.auto import tqdm
import time

# Check GPU
print(f"PyTorch: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
print(f"GPU count: {torch.cuda.device_count()}")
for i in range(torch.cuda.device_count()):
    print(f"  GPU {i}: {torch.cuda.get_device_name(i)}")
```

```python
# ============== CONFIGURATION ==============
# Dataset size options for speed testing
DATASET_OPTIONS = {
    "tiny": 100_000,      # ~5 min/epoch - Debug
    "small": 1_000_000,   # ~30 min/epoch - Quick test
    "medium": 5_000_000,  # ~2.5 hr/epoch - Baseline
    "large": 10_000_000,  # ~5 hr/epoch - Better perf
    "full": None          # ~17 hr/epoch - Best perf
}

# ⚡ CHANGE THIS TO EXPERIMENT ⚡
SELECTED_DATASET = "small"  # Options: tiny, small, medium, large, full
NUM_SAMPLES = DATASET_OPTIONS[SELECTED_DATASET]

# Model config (NanoGPT-style ~10M params)
class ModelConfig:
    n_layer = 6
    n_head = 6
    n_embd = 384
    block_size = 256  # max sequence length
    dropout = 0.1
    bias = False

# Training config
class TrainConfig:
    batch_size = 64          # per GPU
    grad_accum_steps = 4     # effective batch = 64 * 4 * 2 GPUs = 512
    learning_rate = 3e-4
    weight_decay = 0.1
    max_epochs = 3
    warmup_steps = 1000
    eval_interval = 500
    save_interval = 2000
    use_fp16 = True

config = ModelConfig()
train_cfg = TrainConfig()

print(f"Dataset: {SELECTED_DATASET} ({NUM_SAMPLES or 'all'} samples)")
print(f"Model: {config.n_layer}L-{config.n_head}H-{config.n_embd}D")
```

## 2. Custom UCI Tokenizer

```python
class UCITokenizer:
    """Custom tokenizer for UCI chess moves."""
    
    def __init__(self):
        # Build vocabulary from all possible UCI moves
        self.special_tokens = ["<PAD>", "<BOS>", "<EOS>", "<UNK>"]
        self.pad_token_id = 0
        self.bos_token_id = 1
        self.eos_token_id = 2
        self.unk_token_id = 3
        
        # Generate all possible moves
        files = "abcdefgh"
        ranks = "12345678"
        promotions = ["q", "r", "b", "n"]
        
        moves = set()
        for f1 in files:
            for r1 in ranks:
                for f2 in files:
                    for r2 in ranks:
                        if f1 + r1 != f2 + r2:
                            move = f1 + r1 + f2 + r2
                            moves.add(move)
                            # Promotions
                            if r2 in ["1", "8"]:
                                for p in promotions:
                                    moves.add(move + p)
        
        self.moves = sorted(list(moves))
        self.vocab = self.special_tokens + self.moves
        self.token2id = {t: i for i, t in enumerate(self.vocab)}
        self.id2token = {i: t for t, i in self.token2id.items()}
        self.vocab_size = len(self.vocab)
        
    def encode(self, text, add_special=True, max_length=None):
        moves = text.strip().split()
        ids = [self.token2id.get(m, self.unk_token_id) for m in moves]
        if add_special:
            ids = [self.bos_token_id] + ids + [self.eos_token_id]
        if max_length:
            ids = ids[:max_length]
            ids += [self.pad_token_id] * (max_length - len(ids))
        return ids
    
    def decode(self, ids, skip_special=True):
        tokens = [self.id2token.get(i, "<UNK>") for i in ids]
        if skip_special:
            tokens = [t for t in tokens if t not in self.special_tokens]
        return " ".join(tokens)
    
    def save(self, path):
        with open(path, "w") as f:
            json.dump({"vocab": self.vocab}, f)
    
    @classmethod
    def load(cls, path):
        tok = cls()
        with open(path) as f:
            data = json.load(f)
        tok.vocab = data["vocab"]
        tok.token2id = {t: i for i, t in enumerate(tok.vocab)}
        tok.id2token = {i: t for t, i in tok.token2id.items()}
        tok.vocab_size = len(tok.vocab)
        return tok

tokenizer = UCITokenizer()
print(f"Vocabulary size: {tokenizer.vocab_size}")
print(f"Sample encode: {tokenizer.encode('e2e4 e7e5 g1f3', max_length=10)}")
print(f"Sample decode: {tokenizer.decode([1, 1243, 1199, 1427, 2])}")
```

## 3. Dataset Loading

```python
class ChessDataset(Dataset):
    def __init__(self, data, tokenizer, max_length=256):
        self.data = data
        self.tokenizer = tokenizer
        self.max_length = max_length
    
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        transcript = self.data[idx]["Transcript"]
        ids = self.tokenizer.encode(transcript, max_length=self.max_length)
        x = torch.tensor(ids[:-1], dtype=torch.long)
        y = torch.tensor(ids[1:], dtype=torch.long)
        return x, y

print("Loading dataset from HuggingFace...")
dataset = load_dataset("huyvux3005/lichess-uci-dataset-dec2023", split="train")
print(f"Total samples: {len(dataset):,}")

# Sample if needed
if NUM_SAMPLES and NUM_SAMPLES < len(dataset):
    dataset = dataset.shuffle(seed=42).select(range(NUM_SAMPLES))
    print(f"Using {NUM_SAMPLES:,} samples")

# Split
split = dataset.train_test_split(test_size=0.01, seed=42)
train_data = split["train"]
val_data = split["test"]

train_dataset = ChessDataset(train_data, tokenizer, config.block_size)
val_dataset = ChessDataset(val_data, tokenizer, config.block_size)

train_loader = DataLoader(train_dataset, batch_size=train_cfg.batch_size, shuffle=True, num_workers=2, pin_memory=True)
val_loader = DataLoader(val_dataset, batch_size=train_cfg.batch_size, shuffle=False, num_workers=2, pin_memory=True)

print(f"Train: {len(train_dataset):,} | Val: {len(val_dataset):,}")
```

## 4. NanoGPT Model Architecture

```python
class CausalSelfAttention(nn.Module):
    def __init__(self, config):
        super().__init__()
        assert config.n_embd % config.n_head == 0
        self.c_attn = nn.Linear(config.n_embd, 3 * config.n_embd, bias=config.bias)
        self.c_proj = nn.Linear(config.n_embd, config.n_embd, bias=config.bias)
        self.attn_dropout = nn.Dropout(config.dropout)
        self.resid_dropout = nn.Dropout(config.dropout)
        self.n_head = config.n_head
        self.n_embd = config.n_embd
        self.dropout = config.dropout
        self.register_buffer("bias", torch.tril(torch.ones(config.block_size, config.block_size))
                             .view(1, 1, config.block_size, config.block_size))

    def forward(self, x):
        B, T, C = x.size()
        q, k, v = self.c_attn(x).split(self.n_embd, dim=2)
        k = k.view(B, T, self.n_head, C // self.n_head).transpose(1, 2)
        q = q.view(B, T, self.n_head, C // self.n_head).transpose(1, 2)
        v = v.view(B, T, self.n_head, C // self.n_head).transpose(1, 2)
        
        # Use Flash Attention if available
        if hasattr(F, 'scaled_dot_product_attention'):
            y = F.scaled_dot_product_attention(q, k, v, attn_mask=None, 
                                               dropout_p=self.dropout if self.training else 0,
                                               is_causal=True)
        else:
            att = (q @ k.transpose(-2, -1)) * (1.0 / math.sqrt(k.size(-1)))
            att = att.masked_fill(self.bias[:,:,:T,:T] == 0, float('-inf'))
            att = F.softmax(att, dim=-1)
            att = self.attn_dropout(att)
            y = att @ v
        
        y = y.transpose(1, 2).contiguous().view(B, T, C)
        y = self.resid_dropout(self.c_proj(y))
        return y

class MLP(nn.Module):
    def __init__(self, config):
        super().__init__()
        self.c_fc = nn.Linear(config.n_embd, 4 * config.n_embd, bias=config.bias)
        self.gelu = nn.GELU()
        self.c_proj = nn.Linear(4 * config.n_embd, config.n_embd, bias=config.bias)
        self.dropout = nn.Dropout(config.dropout)

    def forward(self, x):
        return self.dropout(self.c_proj(self.gelu(self.c_fc(x))))

class Block(nn.Module):
    def __init__(self, config):
        super().__init__()
        self.ln_1 = nn.LayerNorm(config.n_embd)
        self.attn = CausalSelfAttention(config)
        self.ln_2 = nn.LayerNorm(config.n_embd)
        self.mlp = MLP(config)

    def forward(self, x):
        x = x + self.attn(self.ln_1(x))
        x = x + self.mlp(self.ln_2(x))
        return x

class ChessGPT(nn.Module):
    def __init__(self, config, vocab_size):
        super().__init__()
        self.config = config
        self.transformer = nn.ModuleDict(dict(
            wte = nn.Embedding(vocab_size, config.n_embd),
            wpe = nn.Embedding(config.block_size, config.n_embd),
            drop = nn.Dropout(config.dropout),
            h = nn.ModuleList([Block(config) for _ in range(config.n_layer)]),
            ln_f = nn.LayerNorm(config.n_embd),
        ))
        self.lm_head = nn.Linear(config.n_embd, vocab_size, bias=False)
        self.transformer.wte.weight = self.lm_head.weight  # weight tying
        
        # Init weights
        self.apply(self._init_weights)
        n_params = sum(p.numel() for p in self.parameters())
        print(f"Model parameters: {n_params/1e6:.2f}M")

    def _init_weights(self, module):
        if isinstance(module, nn.Linear):
            torch.nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.bias is not None:
                torch.nn.init.zeros_(module.bias)
        elif isinstance(module, nn.Embedding):
            torch.nn.init.normal_(module.weight, mean=0.0, std=0.02)

    def forward(self, idx, targets=None):
        device = idx.device
        b, t = idx.size()
        pos = torch.arange(0, t, dtype=torch.long, device=device)
        
        tok_emb = self.transformer.wte(idx)
        pos_emb = self.transformer.wpe(pos)
        x = self.transformer.drop(tok_emb + pos_emb)
        
        for block in self.transformer.h:
            x = block(x)
        x = self.transformer.ln_f(x)
        logits = self.lm_head(x)
        
        loss = None
        if targets is not None:
            loss = F.cross_entropy(logits.view(-1, logits.size(-1)), targets.view(-1), ignore_index=0)
        return logits, loss

    @torch.no_grad()
    def generate(self, idx, max_new_tokens, temperature=1.0, top_k=None):
        for _ in range(max_new_tokens):
            idx_cond = idx[:, -self.config.block_size:]
            logits, _ = self(idx_cond)
            logits = logits[:, -1, :] / temperature
            if top_k is not None:
                v, _ = torch.topk(logits, min(top_k, logits.size(-1)))
                logits[logits < v[:, [-1]]] = -float('Inf')
            probs = F.softmax(logits, dim=-1)
            idx_next = torch.multinomial(probs, num_samples=1)
            idx = torch.cat((idx, idx_next), dim=1)
        return idx

# Create model
config.vocab_size = tokenizer.vocab_size
model = ChessGPT(config, tokenizer.vocab_size)

# Multi-GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
if torch.cuda.device_count() > 1:
    print(f"Using {torch.cuda.device_count()} GPUs with DataParallel")
    model = nn.DataParallel(model)
model = model.to(device)
```

## 5. Training Loop

```python
# Optimizer
optimizer = torch.optim.AdamW(model.parameters(), lr=train_cfg.learning_rate, 
                              weight_decay=train_cfg.weight_decay, betas=(0.9, 0.95))

# Scheduler
def get_lr(step, warmup_steps, max_steps, max_lr, min_lr=1e-5):
    if step < warmup_steps:
        return max_lr * step / warmup_steps
    if step > max_steps:
        return min_lr
    decay_ratio = (step - warmup_steps) / (max_steps - warmup_steps)
    return min_lr + 0.5 * (max_lr - min_lr) * (1 + math.cos(math.pi * decay_ratio))

# Scaler for FP16
scaler = GradScaler() if train_cfg.use_fp16 else None

# Training metrics
metrics = {"train_loss": [], "val_loss": [], "tokens_per_sec": []}
```

```python
@torch.no_grad()
def evaluate(model, val_loader, device):
    model.eval()
    losses = []
    for x, y in val_loader:
        x, y = x.to(device), y.to(device)
        with autocast(enabled=train_cfg.use_fp16):
            _, loss = model(x, y)
        losses.append(loss.item())
    model.train()
    return sum(losses) / len(losses)

def train():
    model.train()
    total_steps = len(train_loader) * train_cfg.max_epochs
    step = 0
    
    print(f"\n{'='*60}")
    print(f"Starting training: {total_steps:,} total steps")
    print(f"{'='*60}\n")
    
    for epoch in range(train_cfg.max_epochs):
        pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{train_cfg.max_epochs}")
        epoch_loss = 0
        tokens_processed = 0
        start_time = time.time()
        
        for batch_idx, (x, y) in enumerate(pbar):
            x, y = x.to(device), y.to(device)
            
            # Update learning rate
            lr = get_lr(step, train_cfg.warmup_steps, total_steps, train_cfg.learning_rate)
            for param_group in optimizer.param_groups:
                param_group['lr'] = lr
            
            # Forward + Backward
            with autocast(enabled=train_cfg.use_fp16):
                logits, loss = model(x, y)
                loss = loss / train_cfg.grad_accum_steps
            
            if scaler:
                scaler.scale(loss).backward()
            else:
                loss.backward()
            
            # Gradient accumulation
            if (batch_idx + 1) % train_cfg.grad_accum_steps == 0:
                if scaler:
                    scaler.unscale_(optimizer)
                    torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                    scaler.step(optimizer)
                    scaler.update()
                else:
                    torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                    optimizer.step()
                optimizer.zero_grad(set_to_none=True)
                step += 1
            
            epoch_loss += loss.item() * train_cfg.grad_accum_steps
            tokens_processed += x.numel()
            
            # Logging
            if batch_idx % 50 == 0:
                elapsed = time.time() - start_time
                tokens_per_sec = tokens_processed / elapsed if elapsed > 0 else 0
                pbar.set_postfix({
                    "loss": f"{loss.item() * train_cfg.grad_accum_steps:.4f}",
                    "lr": f"{lr:.2e}",
                    "tok/s": f"{tokens_per_sec:.0f}"
                })
            
            # Evaluation
            if step > 0 and step % train_cfg.eval_interval == 0:
                val_loss = evaluate(model, val_loader, device)
                metrics["val_loss"].append(val_loss)
                print(f"\n  Step {step}: Val Loss = {val_loss:.4f}, PPL = {math.exp(val_loss):.2f}")
            
            # Save checkpoint
            if step > 0 and step % train_cfg.save_interval == 0:
                save_checkpoint(step)
        
        # Epoch summary
        avg_loss = epoch_loss / len(train_loader)
        metrics["train_loss"].append(avg_loss)
        elapsed = time.time() - start_time
        print(f"\nEpoch {epoch+1} complete: Loss={avg_loss:.4f}, Time={elapsed/60:.1f}min")
    
    print("\n" + "="*60)
    print("Training complete!")
    print("="*60)

def save_checkpoint(step):
    raw_model = model.module if hasattr(model, 'module') else model
    checkpoint = {
        "model_state_dict": raw_model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "step": step,
        "config": vars(config),
    }
    torch.save(checkpoint, f"chess_gpt_step_{step}.pt")
    print(f"  Saved checkpoint: chess_gpt_step_{step}.pt")
```

```python
#  START TRAINING
train()
```

## 6. Export to ONNX

```python
import onnx

def export_to_onnx(model, tokenizer, output_path="chess_gpt.onnx"):
    raw_model = model.module if hasattr(model, 'module') else model
    raw_model.eval()
    raw_model = raw_model.to("cpu")
    
    # Dummy input
    dummy_input = torch.randint(0, tokenizer.vocab_size, (1, 64), dtype=torch.long)
    
    # Export
    torch.onnx.export(
        raw_model,
        (dummy_input,),
        output_path,
        input_names=["input_ids"],
        output_names=["logits"],
        dynamic_axes={
            "input_ids": {0: "batch_size", 1: "sequence_length"},
            "logits": {0: "batch_size", 1: "sequence_length"}
        },
        opset_version=14
    )
    
    # Verify
    onnx_model = onnx.load(output_path)
    onnx.checker.check_model(onnx_model)
    print(f" ONNX model saved: {output_path}")
    
    # Move model back to GPU
    raw_model.to(device)
    return output_path

onnx_path = export_to_onnx(model, tokenizer)
print(f"ONNX file size: {os.path.getsize(onnx_path) / 1e6:.2f} MB")
```

## 7. Export to HuggingFace Format

```python
from transformers import GPT2Config, GPT2LMHeadModel, PreTrainedTokenizerFast
from tokenizers import Tokenizer, models, pre_tokenizers

def export_to_huggingface(model, tokenizer, output_dir="chess_gpt_hf"):
    os.makedirs(output_dir, exist_ok=True)
    raw_model = model.module if hasattr(model, 'module') else model
    
    # Create HF config
    hf_config = GPT2Config(
        vocab_size=tokenizer.vocab_size,
        n_positions=config.block_size,
        n_embd=config.n_embd,
        n_layer=config.n_layer,
        n_head=config.n_head,
        bos_token_id=tokenizer.bos_token_id,
        eos_token_id=tokenizer.eos_token_id,
        pad_token_id=tokenizer.pad_token_id,
    )
    
    # Create HF model and load weights
    hf_model = GPT2LMHeadModel(hf_config)
    
    # Map weights
    state_dict = raw_model.state_dict()
    new_state = {}
    
    # Weight mapping
    new_state["transformer.wte.weight"] = state_dict["transformer.wte.weight"]
    new_state["transformer.wpe.weight"] = state_dict["transformer.wpe.weight"]
    new_state["transformer.ln_f.weight"] = state_dict["transformer.ln_f.weight"]
    new_state["transformer.ln_f.bias"] = torch.zeros(config.n_embd)
    new_state["lm_head.weight"] = state_dict["lm_head.weight"]
    
    for i in range(config.n_layer):
        prefix_old = f"transformer.h.{i}"
        prefix_new = f"transformer.h.{i}"
        
        new_state[f"{prefix_new}.ln_1.weight"] = state_dict[f"{prefix_old}.ln_1.weight"]
        new_state[f"{prefix_new}.ln_1.bias"] = torch.zeros(config.n_embd)
        new_state[f"{prefix_new}.ln_2.weight"] = state_dict[f"{prefix_old}.ln_2.weight"]
        new_state[f"{prefix_new}.ln_2.bias"] = torch.zeros(config.n_embd)
        
        new_state[f"{prefix_new}.attn.c_attn.weight"] = state_dict[f"{prefix_old}.attn.c_attn.weight"].t()
        new_state[f"{prefix_new}.attn.c_attn.bias"] = torch.zeros(3 * config.n_embd)
        new_state[f"{prefix_new}.attn.c_proj.weight"] = state_dict[f"{prefix_old}.attn.c_proj.weight"].t()
        new_state[f"{prefix_new}.attn.c_proj.bias"] = torch.zeros(config.n_embd)
        
        new_state[f"{prefix_new}.mlp.c_fc.weight"] = state_dict[f"{prefix_old}.mlp.c_fc.weight"].t()
        new_state[f"{prefix_new}.mlp.c_fc.bias"] = torch.zeros(4 * config.n_embd)
        new_state[f"{prefix_new}.mlp.c_proj.weight"] = state_dict[f"{prefix_old}.mlp.c_proj.weight"].t()
        new_state[f"{prefix_new}.mlp.c_proj.bias"] = torch.zeros(config.n_embd)
    
    hf_model.load_state_dict(new_state, strict=False)
    hf_model.save_pretrained(output_dir)
    
    # Create HF tokenizer
    vocab = {tok: i for i, tok in enumerate(tokenizer.vocab)}
    hf_tokenizer = Tokenizer(models.WordLevel(vocab=vocab, unk_token="<UNK>"))
    hf_tokenizer.pre_tokenizer = pre_tokenizers.Whitespace()
    
    wrapped_tokenizer = PreTrainedTokenizerFast(
        tokenizer_object=hf_tokenizer,
        bos_token="<BOS>",
        eos_token="<EOS>",
        unk_token="<UNK>",
        pad_token="<PAD>"
    )
    wrapped_tokenizer.save_pretrained(output_dir)
    
    print(f" HuggingFace model saved: {output_dir}/")
    return output_dir

hf_dir = export_to_huggingface(model, tokenizer)
print(f"\nFiles in {hf_dir}:")
for f in os.listdir(hf_dir):
    print(f"  - {f}")
```

## 8. Inference & Generation

```python
def generate_moves(model, tokenizer, prompt="e2e4", max_moves=20, temperature=0.8):
    """Generate chess moves continuation."""
    raw_model = model.module if hasattr(model, 'module') else model
    raw_model.eval()
    
    # Encode prompt
    ids = tokenizer.encode(prompt, add_special=True)
    ids = ids[:-1]  # Remove EOS
    x = torch.tensor([ids], dtype=torch.long, device=device)
    
    # Generate
    with torch.no_grad():
        output = raw_model.generate(x, max_new_tokens=max_moves, temperature=temperature, top_k=40)
    
    # Decode
    generated = tokenizer.decode(output[0].tolist())
    return generated

# Test generation
print("\n" + "="*60)
print("Testing Move Generation")
print("="*60)

test_openings = [
    "e2e4",                    # King's Pawn
    "e2e4 e7e5 g1f3",          # Italian/Spanish setup
    "d2d4 d7d5 c2c4",          # Queen's Gambit
    "e2e4 c7c5",               # Sicilian Defense
]

for opening in test_openings:
    print(f"\nPrompt: {opening}")
    continuation = generate_moves(model, tokenizer, opening, max_moves=15)
    print(f"Generated: {continuation}")
```
